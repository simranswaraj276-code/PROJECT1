```python
# Get the size of the system
num_eq = int(input("Enter the number of variables/equations: "))

# Validate input size
if num_eq <= 0:
    print("Invalid size. Must be greater than 0.")
else:
    # Get coefficients for the matrix
    matrix_A = []
    print("\nInput the coefficient matrix:")
    
    for r in range(num_eq):
        curr_row = []
        for c in range(num_eq):
            curr_row.append(float(input(f"Element A[{r + 1}][{c + 1}]: ")))
        matrix_A.append(curr_row)

    # Get constants for the RHS vector
    vector_b = []
    print("\nInput the constant terms (RHS):")
    
    for r in range(num_eq):
        vector_b.append(float(input(f"b[{r + 1}]: ")))

    # Show initial coefficient matrix
    print("\n--- Original Coefficient Matrix ---")
    for r in range(num_eq):
        for c in range(num_eq):
            print(matrix_A[r][c], end="\t")
        print()

    # Show RHS vector
    print("\n--- Constants Vector ---")
    for val in vector_b:
        print(val)

    # Build the augmented matrix
    aug_mat = []
    for r in range(num_eq):
        row_copy = list(matrix_A[r])
        row_copy.append(vector_b[r])
        aug_mat.append(row_copy)

    # Show the augmented matrix
    print("\n--- Initial Augmented Matrix ---")
    for r in range(num_eq):
        for c in range(num_eq + 1):
            print(aug_mat[r][c], end="\t")
        print()

    # Perform Gaussian Elimination with Partial Pivoting
    is_singular = False

    for k in range(num_eq):
        # Locate the best pivot in the current column
        pivot_row = k
        for r in range(k + 1, num_eq):
            if abs(aug_mat[r][k]) > abs(aug_mat[pivot_row][k]):
                pivot_row = r

        # Swap the current row with the pivot row if necessary
        if pivot_row != k:
            aug_mat[k], aug_mat[pivot_row] = aug_mat[pivot_row], aug_mat[k]

        # Guard against division by zero
        if abs(aug_mat[k][k]) < 1e-6:
            is_singular = True
            break

        # Eliminate entries below the pivot
        for r in range(k + 1, num_eq):
            multiplier = aug_mat[r][k] / aug_mat[k][k]
            for c in range(k, num_eq + 1):
                aug_mat[r][c] -= multiplier * aug_mat[k][c]

    # Handle singularity
    if is_singular:
        print("\nFailed: Matrix is singular, has a zero pivot, or lacks a unique solution.")
    else:
        # Show the upper triangular result
        print("\n--- Matrix After Forward Elimination ---")
        for r in range(num_eq):
            for c in range(num_eq + 1):
                print(round(aug_mat[r][c], 4), end="\t")
            print()

        # Execute Back Substitution
        solutions = [0.0 for _ in range(num_eq)]

        for r in range(num_eq - 1, -1, -1):
            sum_val = 0.0
            for c in range(r + 1, num_eq):
                sum_val += aug_mat[r][c] * solutions[c]
            
            solutions[r] = (aug_mat[r][num_eq] - sum_val) / aug_mat[r][r]

        # Output the final calculated variables
        print("\n--- Final Solutions ---")
        for r in range(num_eq):
            print(f"x_{r + 1} = {round(solutions[r], 4)}")

        # Verify the calculations by plugging them back in
        print("\n--- Checking Answers ---")
        for r in range(num_eq):
            calc_lhs = 0.0
            for c in range(num_eq):
                calc_lhs += matrix_A[r][c] * solutions[c]
            
            print(f"Equation {r + 1} -> Calculated LHS: {round(calc_lhs, 4)} | Actual RHS: {vector_b[r]}")



```
